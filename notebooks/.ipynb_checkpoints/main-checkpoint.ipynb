# User choices in scenario, definition of X and Y

```python
#------- Scenario choice
scenario = "ssp245" # ssp119, ssp126, ssp245, ssp370, ssp585

#------- Y definition choice
# Y is the average on the period between min_Y and max_Y
# Its values can be anomalies, compared to the reference period between min_Y_ref and max_Y_ref
anomalie_Y = False
min_Y      = 2091
max_Y      = 2100
min_Y_ref  = 1850
max_Y_ref  = 1900

#------- X definition choice
# X is rather an mean or a trend on a time period between min_X and max_X.
# For mean, we used min_X = 1900
# For trend, we used min_X = 1975
X_choice = "mean" # mean, trend
min_X    = 1980 # 1900 or 1975
max_X    = 2020

#------- Choose if you want to consider the past AMOC in the set of observable variables,
#------- in addition to the 18 observables variables (9 regions for 2 climate variables: the temperature and salinity)
include_AMOC_in_predictors = True

#------- Impose or not to use all the same 17 models, whatever the SSP scenario (except ssp119)
impose_same_17_models = False
```

# Import data and format data into X and Y
There is two cases:
- (univariate) X is only composed by one variable, the past AMOC.
    - The values simulated by the M climate models are stored in X_simu_AMOC.
    - The value observed in the real-world (RAPID) is store in X_obs_AMOC.
- (multivariate) X is composed by multiple variables.
    - The values simulated by the M climate models are stored in X_simu.
    - The value observed in the real-world (RAPID) is store in X_obs.

Y contains the future AMOC average, for each climate model.


```python
[
    X_simu, Y_simu, Y_ref, X_simu_AMOC,
    X_AMOC_mean_1850_1900, final_name_models,
    X_obs, X_obs_AMOC, name_X_AMOC, name_Y,
    mask_perVar, name_perBox, list_idCell_perFeature,
    list_id_box_perFeature, list_id_var_perFeature,
    latitudes, longitudes, middle_cell_perBox, lonlat_lim_perBox, list_name_perFeature
] = construct_data_given_scenario(scenario,
                                 X_choice, anomalie_Y, min_Y, max_Y,
                                 min_Y_ref, max_Y_ref, min_X, max_X,
                                 include_AMOC_in_predictors, return_all=True)
```

```python
if impose_same_17_models:

    target_models = np.array(['ACCESS-CM2', 'ACCESS-ESM1-5', 'CESM2', 'CESM2-WACCM',
           'CNRM-CM6-1', 'CanESM5', 'CanESM5-CanOE', 'FGOALS-f3-L',
           'FGOALS-g3', 'GISS-E2-1-G', 'INM-CM4-8', 'INM-CM5-0',
           'IPSL-CM6A-LR', 'MIROC6', 'MPI-ESM1-2-HR', 'MPI-ESM1-2-LR',
           'MRI-ESM2-0'])

    final_name_models, id_models_to_keep, _ = np.intersect1d(final_name_models, target_models, return_indices=True)

    X_simu                = X_simu[id_models_to_keep]
    Y_simu                = Y_simu[id_models_to_keep]
    Y_ref                 = Y_ref[id_models_to_keep]
    X_simu_AMOC           = X_simu_AMOC[id_models_to_keep]
    X_AMOC_mean_1850_1900 = X_AMOC_mean_1850_1900
    if X_simu.shape[0]!=17:
        raise NameError("There are missing climate models")

```

# Compute the performances of each observational constraint methods
**Warning !** This step takes a lot of time !

The different methods are: the multi-model mean (MMM), weighted average (wAverage), unregularized linear regression (LR), ridge-regularized linear regression (Ridge) and random forest (RF). There are used twice, in both univariate and multivariate cases, i.e. when X is composed by only one or multiple variables.

Variables output:
- Each method's name is stored in "name_methods".
- The guesses for each case and each method are stored in "list_list_predictions".
- The standard deviation error for each case and each method are stored in "list_list_std".
- The mean leave-one out error for each case and each method are stored in "list_list_LOO".
- The weights of the weighted average method are stored in "weights", for the univariate case.

```python
[name_methods, list_list_predictions, list_list_std, list_list_LOO, weights
        ] = performances_methods(X_simu_AMOC, X_obs_AMOC, X_simu, X_obs, Y_simu, return_LOOperFold=True)
```

## Values for the article

```python
name_methods = ["Multi-Model Mean", "Weighted Average", "Linear Regression", "Ridge Regression", "Random Forest"]
id_ridge     = 3

print("How many percent ridge regression makes a better cross-validation error than other methods:")
for id_method in range(len(name_methods)):
    LOO = np.mean(list_list_LOO[1][id_method])
    LOO_ridge = np.mean(list_list_LOO[1][id_ridge])
    print("\t{}: {:.0f}%".format(name_methods[id_method], 100*(LOO-LOO_ridge)/LOO))
    
```

# Decompose the correction induced by the observational constraint made by ridge regression when using multiple observable variables

```python
#------ Compute the coefficients of Ridge regression
# Data are standardised before the ridge regression
scaler    = StandardScaler(with_mean=False)
X_simu_n  = scaler.fit_transform(X_simu)
X_obs_n   = scaler.transform(X_obs.reshape(1,-1))
# Ridge-regularised regression (the parameter is automatically selected using cross-validation)
model     = model_RidgeCV_noisy(Cov_obs=0).fit(X_simu_n, Y_simu)
# Store the resulting linear coefficients
Beta      = model.model.coef_
intercept = model.model.intercept_

# Compute and store the coefficients and correction in a data table 
pd.set_option("display.precision", 2)
data = np.concatenate((Beta.reshape(-1,1),
                       (X_obs_n-np.mean(X_simu_n, axis=0)).reshape(-1,1),
                       (Beta*(X_obs_n-np.mean(X_simu_n, axis=0))).reshape(-1,1)), axis=1)
```

## Group both TOS and SOS influence in each region, to sort by the most important regions

```python
name_region_per_feature = np.array([elt.split('os ')[1] for elt in list_name_perFeature[:-1]])
name_feature_AMOC = list_name_perFeature[-1]

unique_name_regions = np.unique(name_region_per_feature)

list_name_feature_bis = []
data_bis = np.zeros((len(unique_name_regions)+1, data.shape[1]))
id_feature_bis = 0
for name_region in unique_name_regions:
    id_tos_region = np.argwhere('tos '+name_region==np.array(list_name_perFeature))[0][0]
    id_sos_region = np.argwhere('sos '+name_region==np.array(list_name_perFeature))[0][0]
    
    data_bis[id_feature_bis, :] = data[id_tos_region] + data[id_sos_region]
    list_name_feature_bis.append('tos & sos '+name_region)
    id_feature_bis += 1
    
data_bis[-1, :] = data[-1, :]
list_name_feature_bis.append(name_feature_AMOC)
```

# Table of regions influences, sorted by the most important regions

```python
# Indexes
id_sorted_region = np.flip(np.argsort(np.abs(data_bis[:-1, 2])))
indexes = [-1]
for name_region in unique_name_regions[id_sorted_region]:
    id_tos_region = np.argwhere('tos '+name_region==np.array(list_name_perFeature))[0][0]
    id_sos_region = np.argwhere('sos '+name_region==np.array(list_name_perFeature))[0][0]
    indexes.append(id_tos_region)
    indexes.append(id_sos_region)

df = pd.DataFrame(data[indexes], index=np.array(list_name_perFeature)[indexes],
             columns=[r"$\beta$", r"($X_0$-$\mu_{X}$)", r"$\beta$ ($X_0$-$\mu_{X}$)"])


percent_  = 0.75
threshold = np.flip(np.sort(np.abs(data[:, 2])))[np.cumsum(np.flip(np.sort(np.abs(data[:, 2]))))>percent_*np.sum(np.abs(data[:, 2]))][0]
def style_significative(v, props=''):
    if abs(v)> threshold:
        return props
    else:
        return None

s2 = df.style.map(style_significative, props='color:red;', subset=[r"$\beta$ ($X_0$-$\mu_{X}$)"]).format(precision=2)
s2
```

# Constrain the reference 

```python
#------ Compute the coefficients of Ridge regression
# Data are standardised before the ridge regression
scaler    = StandardScaler(with_mean=False)
X_simu_n  = scaler.fit_transform(X_simu)
X_obs_n   = scaler.transform(X_obs.reshape(1,-1))
# Ridge-regularised regression (the parameter is automatically selected using cross-validation)
model     = model_RidgeCV_noisy(Cov_obs=0).fit(X_simu_n, Y_ref)


predictions = model.predict(X_simu_n)
RMSE_constrained_ref = np.sqrt(np.mean(np.square(Y_ref-predictions)))

model.predict(X_obs_n)[0]
```

## Values for the article

```python
# Choose the level of probability
from scipy.stats import t
conf = 0.90
z    = t.interval(conf, np.inf, loc=0, scale=1)[1]
print("{}% probability: +- {:.2f} sigma".format(int(100*conf), z))
```

```python
Y_simu_percent = (Y_simu-Y_ref)/Y_ref
print("AMOC slow down: {:.1f} Sv +- {:.1f} ({} % probability)".format(
    np.mean(Y_simu), z*np.std(Y_simu), int(100*conf)))

ref = np.mean(Y_ref)

Y_ref_constrained = model.predict(X_obs_n)[0]
ref = Y_ref_constrained

print("\nUnconstrained reference ({}-{} average on CMIP6 models): {:.1f} +- {:.1f} Sv".format(
    min_Y_ref, max_Y_ref, np.mean(Y_ref), z*np.std(Y_ref)))

print("Constrained reference ({}-{} average on CMIP6 models): {:.1f} +- {:.1f} Sv".format(
    min_Y_ref, max_Y_ref, ref, z*RMSE_constrained_ref))

print("\nUnconstrained AMOC slow down: {:.1f} % +- {:.1f} % ({} % probability)".format(
    100*(np.mean(Y_simu)-ref)/ref, 100*z*np.std(Y_simu)/ref, int(100*conf)))



# Best guess (in the sense of Leave-One-Out (LOO) cross-validation error)
list_list_LOO_cpy = np.mean(np.array(list_list_LOO), axis=2)
best_guess        = np.array(list_list_predictions)[np.where(list_list_LOO_cpy==np.min(list_list_LOO_cpy))][0]
best_guess_error  = np.array(list_list_std)[np.where(list_list_LOO_cpy==np.min(list_list_LOO_cpy))][0]

print("Constrained AMOC slow down: {:.1f} +- {:.1f} ({} % probability)".format(
    best_guess, z*best_guess_error, int(100*conf)))
print("Constrained AMOC slow down: {:.1f} % +- {:.1f} % ({} % probability)".format(
    100*(best_guess-ref)/ref, 100*z*best_guess_error/ref, int(100*conf)))

print("Reduction of uncertainty: {:.1f} %".format(100*(np.std(Y_simu)-best_guess_error)/np.std(Y_simu)))
print("Stronger AMOC decline by {:.1f} %".format(100*(best_guess-np.mean(Y_simu))/np.mean(Y_simu)))
```

## The constrain using ORAS5 (warning: you have to chose the mean 1980-2020 in the begining of this script)

```python
# Replace the SST features with the values from ORAS5
X_obs_2 = np.copy(X_obs)

mean_1980_2020_SSS_ORAS5_perRegion = np.load("../data/mean_1980_2020_SSS_ORAS5_perRegion.npy")
mean_1980_2020_SST_ORAS5_perRegion = np.load("../data/mean_1980_2020_SST_ORAS5_perRegion.npy")

for i in range(len(mean_1980_2020_SSS_ORAS5_perRegion)):
    X_obs_2[2*i] = mean_1980_2020_SST_ORAS5_perRegion[i]
    X_obs_2[2*i+1] = mean_1980_2020_SSS_ORAS5_perRegion[i]
```

```python
scaler    = StandardScaler(with_mean=False)
X_simu_n  = scaler.fit_transform(X_simu)
X_obs_2_n = scaler.transform(X_obs_2.reshape(1,-1))
model     = model_RidgeCV_noisy(Cov_obs=0).fit(X_simu_n, Y_simu)

prediction = model.predict(X_obs_2_n)[0]
predictions = model.predict(X_simu_n)
RMSE_constrained = np.sqrt(np.mean(np.square(Y_simu-predictions)))

print("Constrained AMOC slow down: {:.1f} +- {:.1f} ({} % probability)".format(
    prediction, z*RMSE_constrained, int(100*conf)))
```
